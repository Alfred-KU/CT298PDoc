```python
import csv
import pandas as pd
import yfinance as yf

# 1. Define your list of stock tickers correctly
column_data = []
with open("stocklisttop30.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  next(reader)  # Skip header if present
  for row in reader:
    if row:  # Ensure row is not empty
      column_data.append(row[0])

stock_list = column_data  # Assign the populated list directly

# 2. Loop through each stock in the list
for stock in stock_list:
  try:
    # Download the data
    hkstock = yf.download(
        stock,
        start="2023-01-01",
        end="2026-01-01",
        auto_adjust=True,
        progress=False,
    )

    # Drop the Ticker level if it exists in the MultiIndex columns
    if isinstance(hkstock.columns, pd.MultiIndex):
      hkstock = hkstock.droplevel("Ticker", axis=1)

    close_df = hkstock[["Close"]]

    # 3. Create a dynamic filename by stripping the .HK and adding the date
    clean_name = stock.replace(".HK", "")
    filename = f"{clean_name}.HK260101.csv"

    # Save to CSV
    close_df.to_csv(filename)
    print(f"Successfully saved: {filename}")

  except Exception as e:
    print(f"Error downloading {stock}: {e}")
```

```python
import csv
import pandas as pd
import yfinance as yf

# 1. Define your list of stock tickers correctly
column_data = []
with open("stocklisttop30.csv", mode="r", newline="", encoding="utf-8") as file:
  reader = csv.reader(file)
  next(reader)  # Skip header if present
  for row in reader:
    if row:  # Ensure row is not empty
      column_data.append(row[0])

stock_list = column_data  # Assign the populated list directly

# 2. Loop through each stock in the list
for stock in stock_list:
  try:
    # Download the data
    hkstock = yf.download(
        stock,
        start="2023-07-01",
        end="2026-07-01",
        auto_adjust=True,
        progress=False,
    )

    # Drop the Ticker level if it exists in the MultiIndex columns
    if isinstance(hkstock.columns, pd.MultiIndex):
      hkstock = hkstock.droplevel("Ticker", axis=1)

    close_df = hkstock[["Close"]]

    # 3. Create a dynamic filename by stripping the .HK and adding the date
    clean_name = stock.replace(".HK", "")
    filename = f"{clean_name}.HK260701.csv"

    # Save to CSV
    close_df.to_csv(filename)
    print(f"Successfully saved: {filename}")

  except Exception as e:
    print(f"Error downloading {stock}: {e}")
```
